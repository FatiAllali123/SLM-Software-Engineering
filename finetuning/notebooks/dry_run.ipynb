4: load the model and data, and run three checks

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "0"
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only
import torch, json, time
from datasets import Dataset
from trl import SFTTrainer, SFTConfig

CFG = {"model": "Qwen/Qwen2.5-Coder-1.5B-Instruct", "max_seq_length": 2048, "r": 16, "lora_alpha": 16,
       "lora_dropout": 0, "lr": 2e-4, "per_device_batch": 2, "grad_accum": 8, "seed": 42, "dry_steps": 30}

model, tok = FastLanguageModel.from_pretrained(CFG["model"], max_seq_length=CFG["max_seq_length"],
                                               load_in_4bit=True, dtype=None)
model = FastLanguageModel.get_peft_model(
    model, r=CFG["r"], lora_alpha=CFG["lora_alpha"], lora_dropout=CFG["lora_dropout"],
    target_modules=["q_proj","k_proj","v_proj","o_proj","gate_proj","up_proj","down_proj"],
    use_gradient_checkpointing="unsloth", random_state=CFG["seed"])

def load(path):
    rows = [json.loads(l) for l in open(path, encoding="utf-8")]
    return Dataset.from_list([{"text": tok.apply_chat_template(r["messages"], tokenize=False)} for r in rows])
train_ds, val_ds = load("/kaggle/working/ft_data/train.jsonl"), load("/kaggle/working/ft_data/val.jsonl")
assert train_ds[0]["text"].startswith("<|im_start|>system\nYou are Qwen"), "template mismatch with benchmark"
print("train/val:", len(train_ds), len(val_ds))

trainer = SFTTrainer(
    model=model, processing_class=tok, train_dataset=train_ds, eval_dataset=val_ds.select(range(100)),
    args=SFTConfig(
        output_dir="/kaggle/working/out_dry", per_device_train_batch_size=CFG["per_device_batch"],
        gradient_accumulation_steps=CFG["grad_accum"], per_device_eval_batch_size=2,
        max_steps=CFG["dry_steps"], warmup_steps=5, learning_rate=CFG["lr"], lr_scheduler_type="linear",
        optim="adamw_8bit", weight_decay=0.01, fp16=not is_bfloat16_supported(), bf16=is_bfloat16_supported(),
        logging_steps=1, eval_strategy="no", save_strategy="no", seed=CFG["seed"], report_to="none",
        max_length=CFG["max_seq_length"], dataset_text_field="text", packing=False, dataset_num_proc=1))
trainer = train_on_responses_only(trainer, instruction_part="<|im_start|>user\n",
                                  response_part="<|im_start|>assistant\n")

# CHECK 1: how the training example ends (watch for a doubled end-of-turn token)
ex = trainer.train_dataset[0]
print("last tokens:", repr(tok.decode(ex["input_ids"][-8:])))
# CHECK 2: only the answer should be trained on (the labels that are not -100)
lab = [t for t, l in zip(ex["input_ids"], ex["labels"]) if l != -100]
print("trained-on text starts:", repr(tok.decode(lab[:60])))
print("trained-on tokens:", len(lab), "of", len(ex["input_ids"]))
```

5: val loss before, 30-step dry run, val loss after

```python
import statistics
before = trainer.evaluate()["eval_loss"]; print("val loss before:", before)
torch.cuda.reset_peak_memory_stats(); t0 = time.time()
out = trainer.train()
wall = time.time() - t0
after = trainer.evaluate()["eval_loss"]; print("val loss after:", after)

losses = [h["loss"] for h in trainer.state.log_history if "loss" in h]
sec_step = out.metrics["train_runtime"] / CFG["dry_steps"]
eff = CFG["per_device_batch"] * CFG["grad_accum"]
print(f"train loss first5 mean {statistics.mean(losses[:5]):.3f} | last5 mean {statistics.mean(losses[-5:]):.3f}")
print(f"{sec_step:.1f} s/step | peak GPU mem {torch.cuda.max_memory_allocated()/1e9:.2f} GB | wall {wall:.0f}s")
print(f"EXTRAPOLATION (estimate, not measured): 1 epoch = {6000//eff} steps = {6000//eff*sec_step/3600:.1f} h")
```

6: save the report and the dry-run adapter

```python
from importlib.metadata import version
rep = {"cfg": CFG, "gpu": torch.cuda.get_device_name(0), "val_loss_before": before, "val_loss_after": after,
       "train_losses": losses, "sec_per_step": sec_step, "peak_gpu_gb": round(torch.cuda.max_memory_allocated()/1e9, 2),
       "versions": {k: version(k) for k in ["unsloth","unsloth_zoo","trl","transformers","peft","torch","bitsandbytes","datasets"]},
       "data_sha256": EXPECTED}
json.dump(rep, open("/kaggle/working/dry_run_report.json", "w"), indent=2)
model.save_pretrained("/kaggle/working/adapter_dry")
tok.save_pretrained("/kaggle/working/adapter_dry")
print("saved")
```
